## 2. ¿Qué es un Análisis de Portafolio de Inversión? 📚

La evaluación de un portafolio de inversión es el procedimiento mediante el cual se examinan y analizan todas las inversiones que tiene una persona o una empresa (su "portafolio") para determinar si funcionan bien en conjunto y si contribuyen a lograr los objetivos financieros establecidos.

**¿Por qué es importante?**

El análisis de portafolio sirve para responder preguntas como:

¿Estoy ganando lo suficiente con mis inversiones?

¿Estoy asumiendo demasiado riesgo?

¿Tengo una buena mezcla de diferentes tipos de inversiones?

¿Mis inversiones están alineadas con mis metas, como ahorrar para la jubilación o comprar una casa?

El análisis busca encontrar el equilibrio adecuado entre el rendimiento (ganancias) y el riesgo (posibilidad de perder dinero), asegurando que tus inversiones estén bien diversificadas y alineadas con tus metas personales.

Durante este análisis, se observa:

La diversificación: Es decir, si tienes diferentes tipos de activos para reducir el riesgo de pérdidas grandes.

El rendimiento: Cuánto han crecido o disminuido tus inversiones en un periodo de tiempo.

El riesgo: Qué tan probable es que pierdas dinero o que el valor de tus inversiones cambie mucho.

El análisis de portafolio permite encontrar un punto medio:

“No poner todos los huevos en la misma canasta”.

- *El Valor Intrínseco*:

Esta forma de análisis se centra en evaluar el valor real o intrínseco de los activos individuales que componen el portafolio (como acciones o bonos). La premisa es que el mercado puede subvalorar o sobrevalorar un activo a corto plazo, pero a la larga, el precio se ajustará a su valor fundamental.

**Para una acción:** Se analizan minuciosamente los estados financieros de la empresa (balances, cuentas de resultados), su modelo de negocio, la calidad de su gestión, las perspectivas del sector industrial al que pertenece y las condiciones macroeconómicas generales. Se utilizan ratios y/o indicadore como el Precio/Beneficio (P/E), Deuda/Patrimonio, y el Retorno sobre el Patrimonio (ROE).

**Para el portafolio:**
 Se consolida esta información para todas las empresas en las que se ha invertido. Se busca determinar si el portafolio en su conjunto está compuesto por empresas sólidas y con buen potencial de crecimiento a largo plazo.

 **¿Por qué es Importante el Valor Intrínseco en una Inversión?**

 Imagina que vas a comprar una casa. ¿Pagas el precio que el vendedor pide sin más, o intentas averiguar cuánto realmente vale esa casa basándote en su tamaño, ubicación, estado y el mercado inmobiliario local? El valor intrínseco es ese "precio real" para cualquier activo de inversión.

 **La importancia del valor intrínseco radica en:**

 **Es el Ancla contra la Irracionalidad del Mercado:** Los precios de mercado fluctúan constantemente debido al miedo, la codicia, las noticias y la especulación (lo que Ben Graham, el maestro de Warren Buffett, llamaba "el señor Mercado", un personaje con cambios de humor impredecibles). El valor intrínseco sirve como una brújula que te indica si el mercado está sobrevalorando ("caro") o infravalorando ("barato") un activo, permitiéndote tomar decisiones racionales en lugar de emocionales.

 **Ayuda a Identificar Oportunidades de Verdad:** Te permite encontrar "gangas" o joyas ocultas que el resto del mercado puede estar ignorando temporalmente. Sin entender el valor intrínseco, es fácil dejarse llevar por la moda del momento y comprar activos que ya están sobrevalorados.

- *Diversificación y correlación*:

**¿Qué es la diversificación?**

La diversificación significa no poner todos los huevos en la misma canasta.
En inversiones, quiere decir repartir el dinero entre diferentes activos (acciones, bonos, fondos, sectores, países, monedas, etc.) para reducir el riesgo total sin necesariamente reducir la rentabilidad esperada.

En palabras parroquiales, “Si un activo baja, otro puede subir o mantenerse estable, equilibrando las pérdidas.”

**¿Qué es la correlación?**

La correlación mide cómo se mueven dos activos en relación uno con otro.
Su valor va de –1 a +1:

*Cuando el valor de la correlación es positivo, es decir +1 indica que se mueven igual y por ende no hay diversificación, quiere decir que ambos suben y bajan juntos.*

*Cuando el valor de la correlación es 0 no hay relación y la diversificación es buena*

*Por el contrario cuando el valor de esta es negativo en este caso -1 significa que se mueven en sentido opuesto y esto permite que haya una excelente diversificación (uno sube cuando el otro baja).*

**¿Por qué es recomendable diversificar y tener correlación baja?**

**Reducción del riesgo total:** Aunque cada activo tenga su propio riesgo, al combinarlos con baja correlación, las subidas y bajadas se compensan y el riesgo combinado del portafolio disminuye.

**Mejor relación riesgo–rendimiento:** La diversificación permite obtener el mismo rendimiento esperado con menor riesgo, o un rendimiento mayor para el mismo nivel de riesgo.
Esto se refleja en la frontera eficiente de Markowitz, que muestra las combinaciones óptimas de activos.

"Cuanto más baja la correlación, mayor el beneficio de diversificación y menor la volatilidad del portafolio."


- *Análisis de rentabilidad y riesgo.*:

El análisis de rentabilidad y riesgo es el proceso de evaluar qué tan rentable es una inversión y qué tanto riesgo implica.
Es decir, sirve para medir la relación entre lo que se puede ganar y lo que se puede perder.

Para entenderlo mejor,
“Toda inversión tiene dos caras: cuánto puedes ganar (rentabilidad) y cuánto podrías perder o cuánto puede variar (riesgo).”

El análisis busca equilibrar esas dos fuerzas para tomar decisiones inteligentes.

**¿Qué es la Rentabilidad?**

La rentabilidad (o rendimiento) mide cuánto ganas o pierdes en una inversión durante un período de tiempo.
Se expresa como porcentaje respecto al dinero invertido.


$$
\text{Rentabilidad} = \frac{\text{Valor final} - \text{Valor inicial}}{\text{Valor inicial}} \times 100
$$


**¿Qué es el Riesgo?**

El riesgo mide la incertidumbre sobre los resultados futuros de una inversión, o la probabilidad de no obtener lo esperado.

En resumidas cuentas

“Riesgo es qué tanto pueden variar los resultados respecto al promedio.”

**Indicador más común:**

Desviación estándar (volatilidad): mide cuánto se alejan los rendimientos de su promedio.

Alta desviación estándar → más riesgo (rendimientos muy variables).

Baja desviación estándar → más estabilidad.

**¿Por qué se analiza rentabilidad y riesgo juntos?**

Porque no sirve de nada ganar mucho si puedes perderlo todo, y tampoco sirve estar completamente seguro si no ganas nada.
